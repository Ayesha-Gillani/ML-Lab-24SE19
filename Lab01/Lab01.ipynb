# Task 6: Difference Between a Tuple and a List

A list is mutable, which means we can change its elements after creating it. A tuple is immutable, which means we cannot change its elements after creating it. Lists use square brackets [], while tuples usually use parentheses ().

# Task 7: Extract the Domain from an Email Address

```python
# Task 7: Extract the domain from an email address
def get_domain(email):
    return email.split("@")[1]

print("Q7:", get_domain("user@domain.com"))
```

# Task 8: Check Whether a String Contains Dog

```python
# Task 8: Check whether a string contains dog
def find_dog(text):
    return "dog" in text.lower()

print("Q8 - Test 1:", find_dog("Is there a dog here?"))
print("Q8 - Test 2:", find_dog("I LOVE DOGS"))
print("Q8 - Test 3:", find_dog("There is a cat here."))
```

# Task 9: Count Occurrences of Dog in a String

```python
# Task 9: Count occurrences of dog in a string
def count_dog(text):
    return text.lower().count("dog")

print("Q9 - Test 1:", count_dog("dog dog dog"))
print("Q9 - Test 2:", count_dog("Dog and DOG and dog"))
print("Q9 - Test 3:", count_dog("There is a cat here."))
```

# Task 10: Calculate a Speeding Ticket

```python
# Task 10: Calculate a speeding ticket
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"


print("Q10 - Test 1:", caught_speeding(60, False))
print("Q10 - Test 2:", caught_speeding(65, True))
print("Q10 - Test 3:", caught_speeding(75, False))
print("Q10 - Test 4:", caught_speeding(85, True))
print("Q10 - Test 5:", caught_speeding(90, False))
print("Q10 - Test 6:", caught_speeding(85, False))
```
